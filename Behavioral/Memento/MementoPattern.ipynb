```python
from MementoPattern import * 
from PIL import Image, ImageEnhance, ImageDraw, ImageFilter, ImageFont
```

```python
stk_img = Image.open('./img/Kanoon.jpg').resize((125,125))
sticker1 = Sticker(stk_img)

careTakerSticker1 = CareTaker()
careTakerSticker1.addMemento(sticker1.save())

# display(sticker1.getImage(),sticker1.getMask())
display(sticker1.makeSticker())
```

```python
editStickerImg = sticker1.getImage().copy()
draw = ImageDraw.Draw(editStickerImg)
text_position = (65, 50)
text = '///'
draw.text(text_position, text)
sticker1.setImage(editStickerImg)

careTakerSticker1.addMemento(sticker1.save())

display(sticker1.makeSticker())
```

```python
enhancer = ImageEnhance.Brightness(editStickerImg)
factor = 1.5
editStickerImg = enhancer.enhance(factor)
sticker1.setImage(editStickerImg)

careTakerSticker1.addMemento(sticker1.save())

display(sticker1.makeSticker())
```

```python
new_mask1 = Image.open('./img/heart_mask.jpg').resize(sticker1.getImage().size)
sticker1.setMask(new_mask1)

display(sticker1.makeSticker())
```

```python
careTakerSticker1.undo()
careTakerSticker1.undo()
display(sticker1.makeSticker())
```

```python
sticker1.setMask(new_mask1)

careTakerSticker1.addMemento(sticker1.save())

display(sticker1.makeSticker())
```

```python
img = Image.open('./img/KafaeAndKati.jpg').resize((250,250))
pic1 = Picture(img)

careTakerPic1 = CareTaker()
careTakerPic1.addMemento(pic1.save())
display(pic1.getImage())
```

```python
editPicture = pic1.getImage().copy()
editPicture = editPicture.convert('L')
pic1.setImage(editPicture)

display(pic1.getImage())
```

```python
careTakerPic1.undo()
display(pic1.getImage())
```

```python
editPicture = pic1.getImage().copy()
enhancer = ImageEnhance.Brightness(editPicture)
factor = 1.1
editPicture = enhancer.enhance(factor)
pic1.setImage(editPicture)

careTakerPic1.addMemento(pic1.save())

display(pic1.getImage())
```

```python
editPicture.paste(sticker1.getImage(), (100,120), sticker1.getMask())
pic1.setImage(editPicture)

careTakerPic1.addMemento(pic1.save())

display(pic1.getImage())
```

```python
editPicture.paste(sticker2.getImage(), (180,10), sticker2.getMask())
pic1.setImage(editPicture)

display(pic1.getImage())
```

```python
careTakerPic1.undo()
careTakerPic1.undo()
display(pic1.getImage())
```

```python
editPicture = pic1.getImage()
editPicture.paste(sticker1.getImage(), (130,130), sticker1.getMask().filter(ImageFilter.GaussianBlur(3)))

careTakerPic1.addMemento(pic1.save())

editPicture.paste(sticker2.getImage(), (180,2), sticker2.getMask())
pic1.setImage(editPicture)

careTakerPic1.addMemento(pic1.save())

display(pic1.getImage())
```

```python
draw = ImageDraw.Draw(editPicture)
font = ImageFont.truetype(font='BLK BangLi-Ko-Sa-Na Regular.ttf',size=30)
text_position = (50, 10)
text_color = (255, 255, 0)
text = "สวัสดีวันจันทร์"
draw.text(text_position, text, text_color, font)
pic1.setImage(editPicture)

careTakerPic1.addMemento(pic1.save())

display(pic1.getImage())
```
