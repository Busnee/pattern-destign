```python
size = (250,250)
```

```python
from abc import ABC, abstractmethod
from PIL import Image

# Interface
class PaintSrategy(ABC):
    @abstractmethod
    def paint(pic: Image) -> Image:
        pass
```

```python
class ChalkPaint(PaintSrategy):
    def __init__(self):
        super().__init__()
    def paint(pic: Image) -> Image:
        return Image.open('./img/pudding_chalk.png').resize(size)
class ColoredPencil(PaintSrategy):
    def __init__(self):
        super().__init__()
    def paint(pic: Image) -> Image:
        return Image.open('./img/pudding_coloredpencil.png').resize(size)
class WaterColor(PaintSrategy):
    def __init__(self):
        super().__init__()
    def paint(pic: Image) -> Image:
        return Image.open('./img/pudding_watercolor.png').resize(size)
class PosterColor(PaintSrategy):
    def __init__(self):
        super().__init__()
    def paint(pic: Image) -> Image:
        return Image.open('./img/pudding_poster.png').resize(size)
```

```python
class DrawingPicture:
    __strategy: PaintSrategy
    __pic: Image
    def __init__(self):
        self.__pic = Image.open('./img/pudding_lineart.png').resize(size)
    def setStrategy(self, s: PaintSrategy):
        self.__strategy = s
    def paintPicture(self) -> Image:
        return self.__strategy.paint()
    def getPic(self):
        return self.__pic
```

```python
drawing = DrawingPicture()
display(drawing.getPic())
```

```python
drawing.setStrategy(ChalkPaint())
display(drawing.paintPicture())
```

```python
drawing.setStrategy(ColoredPencil())
display(drawing.paintPicture())
```

```python
drawing.setStrategy(WaterColor())
display(drawing.paintPicture())
```

```python
drawing.setStrategy(PosterColor())
display(drawing.paintPicture())
```
